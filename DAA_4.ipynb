```python
# Implementation and Time analysis of factorial program using iterative and recursive method
def factorial_iterative(n):
    fact = 1

    for i in range(1, n + 1):
        fact = fact * i

    return fact


# Factorial using Recursive Method
def factorial_recursive(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial_recursive(n - 1)


# Input
n = int(input("Enter a number: "))

# Iterative Method
result_iterative = factorial_iterative(n)

# Recursive Method
result_recursive = factorial_recursive(n)

# Output
print("\nFactorial using Iterative Method:", result_iterative)
print("Factorial using Recursive Method:", result_recursive)

# Time Complexity
print("\nTime Complexity:")
print("Iterative Method: O(n)")
print("Recursive Method: O(n)")

# Space Complexity
print("\nSpace Complexity:")
print("Iterative Method: O(1)")
print("Recursive Method: O(n)")
```
